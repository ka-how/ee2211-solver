## 1. Inputs
Choose `Xw = y` or `wX = y`. For Xw = y, X and y need matching rows.
For wX = y, they need matching columns; for example use X = `1 1 1; 1 2 3`
and y = `2 3 4`. Include the bias column or row yourself when needed.

An underdetermined system returns a minimum-norm solution. Inconsistent systems
return a least-squares fit. Multiple target columns or rows are supported.

```python
equation = 'Xw = y'  # Xw = y | wX = y; include bias entries yourself.
X = '1 1; 1 2; 1 3'  # Numeric coefficient matrix; semicolons separate rows.
y = '2; 3; 4'  # Targets; rows match X for Xw=y, columns match for wX=y.
```

```python
from notebook_tools import solve_linear_system

X, y = matrix(X), matrix(y)
solution = solve_linear_system(X, y, equation)
show_metrics({key: solution[key] for key in ("system", "status", "equations", "unknowns", "rank_coefficients", "rank_augmented")})
show("solve_X", X, cache=cache, source=equation, kind="input")
show("solve_y", y, cache=cache, source=equation, kind="input")
for key in ("w", "predicted", "residual", "mse", "standard_error"):
    show("solve_" + key, solution[key], cache=cache, source=equation)
```

The residual is predicted − observed. As on the website, `standard_error`
is sqrt(MSE), the root mean squared residual per target, with no degrees-of-freedom
correction. The default example gives w = [1, 1]ᵀ and negligible residuals.

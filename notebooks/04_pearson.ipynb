## 1. Inputs
Use rows for observations and columns for variables. Matching single-row vectors
are treated as observation vectors. If rows do not match but columns do, both
arrays are transposed, matching the website. Correlation requires at least two
observations and is undefined for a constant variable.

The output has one row per X variable and one column per Y variable. Means,
variances, standard deviations, and covariance use population conventions.

```python
X = '1 1; 2 1; 3 2; 4 2'  # Numeric matrix: observation rows, X-variable columns.
Y = '2 1; 4 1; 6 2; 8 2'  # Same observation count as X; Y-variable columns.
```

```python
from pearson_correlation import pearson_correlation_details

pearson = pearson_correlation_details(matrix(X), matrix(Y))
show_metrics({key: pearson[key] for key in ("observations", "x_variables", "y_variables", "orientation")})
show("pearson", pearson["pearson"], cache=cache, source="Pearson correlation")
display(pd.DataFrame(pearson["pearson"],
    index=[f"X{i+1}" for i in range(pearson["x_variables"])],
    columns=[f"Y{i+1}" for i in range(pearson["y_variables"])]))
```

```python
for key in ("meanX", "stdX", "varX", "meanY", "stdY", "varY", "covariance"):
    show(key, pearson[key])
```
